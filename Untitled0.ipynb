## Linear Regression by Pytorch


````python
# 建立data car price
# 一般建立數據或是從csv都取 會先轉成ndarray 調整資料型態(ndarray的特性只能同一資料型態) 調整形狀
car_price_array = [3, 4, 5, 6, 7, 8, 9]
car_price_np = np.array(car_price_array, dtype = np.float32)
car_price_np = car_price_np.reshape(-1, 1)
car_price_tensor = Variable(torch.from_numpy(car_price_np))

# 建立car sell
num_of_car_sell_array = [ 7.5, 7, 6.5, 6.0, 5.5, 5.0, 4.5]
num_of_car_sell_np = np.array(num_of_car_sell_array, dtype = np.float32)
num_of_car_sell_np = num_of_car_sell_np.reshape(-1, 1)
num_of_car_sell_tensor = Variable(torch.from_numpy(num_of_car_sell_np))

# 輸出圖表
import matplotlib.pyplot as plt
plt.scatter(car_price_tensor, num_of_car_sell_tensor)
plt.xlabel("car price")
plt.ylabel("number of car sell")
plt.title("car price vs. number of car sell")
plt.show()
````

Steps of Linear Regression

* create LinearRegression class

* define model from this LinearRegression class

* MSE: Mean squared error

* Optimization (SGD:stochastic gradient descent)

* Backpropagation

* Prediction

````python
import torch.nn as nn

# create LinearRegression
class LinearRegression (nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.linear = nn.Linear(input_size, output_size)

    def forward(self, X):
        return self.linear(X)

# define model
input_dim = 1
output_dim = 1
model = LinearRegression(input_dim, output_dim)

#Lossfunction
mse = nn.MSELoss()

#Optimization
learning_rate = 0.02
optimizer = torch.optim.SGD(model.parameters(), lr = learning_rate)

#Training
loss_list = []
iteration_number = 1001
for iteration in range(iteration_number):

    #optimizer
    optimizer.zero_grad()
    #forward
    output = model(car_price_tensor)
    #loss
    loss = mse(output, num_of_car_sell_tensor)
    #backward
    loss.backward()
    #updating parameters based on gradients
    optimizer.step()
    #store loss
    loss_list.append(loss.item())

    #print loss
    if iteration % 100 == 0:
        print(f'epoch:{iteration}, loss:{loss.item()}')

plt.plot(range(iteration_number), loss_list)
plt.xlabel("number of iterations")
plt.ylabel("Loss")
plt.show()
````

預測

````python
# 預測後需將預測值 轉為numpy
# 在另一個檔案中 使用 with torch.no_grad():就不使用detach這樣可以不追蹤梯度
pred = model(car_price_tensor).detach().numpy() # .detach() 不在繼續追蹤梯度
print(pred)
# tensor 需先轉成numpy 可接受（list numpy series）
plt.scatter(car_price_array, num_of_car_sell_array, label = 'original data', color = 'red')
plt.scatter(car_price_array, pred, label = 'predict data', color = 'blue')

plt.legend() # 顯示小標 配合label
plt.xlabel('car price')
plt.ylabel('number of sell')
plt.title('original vs predicted value')
plt.show()
````

### Logistic Regression

1. Import Library

2. Prepare Dataset

3. Create Model

4. Model

5. Loss

6. Optimizer

7. traning the model

8. Prediction

````python
import torch
import torch.nn as nn
from torch.autograd import Variable
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
````

csv檔案第一column是這筆資料的0~9label

後面所有的column都是一個pixel的數值 是一個256色階的灰階

````python
train = pd.read_csv('/content/drive/MyDrive/ML/train.csv', dtype = np.float32)

# pd轉為numpy
features_numpy = train.loc[:, train.columns != 'label'].to_numpy()/255 #正規化輸入feature
targets_numpy = train['label'].to_numpy()
#分割驗證集
features_train, features_val, targets_train, targets_val = train_test_split(features_numpy, targets_numpy, test_size=0.2, random_state=0)
#numpy轉為tensor
featurestrain = torch.from_numpy(features_train)
targetstrain = torch.from_numpy(targets_train).type(torch.LongTensor)
featuresval = torch.from_numpy(features_val)
targetsval = torch.from_numpy(targets_val).type(torch.LongTensor)

batch_size = 100
epochs = 30
# DataLoader 前需透過TensorDataSet將X, y一個一個綁在一起
train = TensorDataset(featurestrain, targetstrain)
val = TensorDataset(featuresval, targetsval)
# DataLoader
train_loader = DataLoader(train, batch_size=batch_size, shuffle=False)
val_loader = DataLoader(val, batch_size=batch_size, shuffle=False)

plt.imshow(features_numpy[10].reshape(28, 28))
plt.title(str(targets_numpy[10]))
plt.show()
````

create model

````python
class LogisticRegression(nn.Module):
    def __init__(self, input_dim, output_dim):
        super().__init__()
        self.linear = nn.Linear(input_dim, output_dim)
    def forward(self, X):
        return self.linear(X)

input_dim = 28 * 28
output_dim = 10

model = LogisticRegression(input_dim, output_dim)

error = nn.CrossEntropyLoss()

learning_rate = 0.001
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
````

training

````python
count = 0
train_loss_list = []
val_loss_list = []
for epoch in range(epochs):
    total = 0
    correct = 0
    for images, labels in train_loader:

        train = Variable(images.reshape(-1, 28*28)) # batch_size, 一串特徵向量
        labels = Variable(labels)

        optimizer.zero_grad()

        outputs = model(train)

        loss = error(outputs, labels)

        loss.backward()

        optimizer.step()

        count += 1

        if count % 100 == 0:
            total += len(labels)
            _, predicted = torch.max(outputs.data, axis=1) #outputs回傳的是每個類別的機率透過max找到最大的值 回傳最大數字,以及index
            correct += (predicted == labels).sum()
            print(f"iteration:{count}, loss:{loss.item()}, accuracy:{100*(correct/total)}%")
        if count % 336 == 0:
            train_loss_list.append(loss.item())


    #test
    val_loss = 0
    val_total = 0
    val_correct = 0
    with torch.no_grad():
        for images, labels in val_loader:
            val_total += len(labels)
            outputs = model(images)
            val_loss += error(outputs, labels).item()
            _, predicted = torch.max(outputs.data, axis=1)
            val_correct += (predicted == labels).sum()
    avg_val_loss = val_loss / len(val_loader)
    val_acc = 100 * (val_correct / val_total)
    val_loss_list.append(avg_val_loss)
    print(f"val:epoch{epoch}, loss:{avg_val_loss}, acc:{val_acc}%")
print(train_loss_list)
print(val_loss_list)
````

Training 時每個 batch 的 loss 是要用來更新權重的

	•	每個 batch 都會：

	1.	forward 得到 loss

	2.	backward 做梯度

	3.	optimizer.step() 更新權重

👉 所以你必須要「即時」知道這個 batch 的 loss，用來訓練。

📌 這些 loss 不是拿來平均，是訓練時必要的中間值。

⸻

🔍 Validation 時你只是想知道整體表現，不需要更新權重，並且一個epoch才計算一次

	•	每個 batch：

	•	forward 得到 loss

	•	不做 backward、也不更新參數

	•	所以只是「蒐集」所有 batch 的 loss，最後平均一下，作為一個評估指標

````python
plt.plot(range(epochs), train_loss_list, label='training_loss', color = 'red')
plt.plot(range(epochs), val_loss_list, label='validation_loss', color = 'blue')
plt.legend()
plt.xlabel('epoch')
plt.ylabel('loss')
plt.title('loss')
plt.show()
````

###ANN


````python
import torch
import torch.nn as nn
from torch.autograd import Variable
from torch.utils.data import DataLoader, TensorDataset
import torch.nn.functional as F
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
````

````python
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("cuda is available.")
````

````python
df = pd.read_csv("/content/drive/MyDrive/ML/train.csv", dtype='float32')

features_np = df.loc[:, df.columns != 'label'].to_numpy()/255
targets_np = df['label'].to_numpy()

features_train, features_val, targets_train, targets_val = train_test_split(features_np, targets_np, test_size=0.2, random_state=0)

train_X = torch.from_numpy(features_train)
train_y = torch.from_numpy(targets_train).type(torch.LongTensor)
val_X = torch.from_numpy(features_val)
val_y = torch.from_numpy(targets_val).type(torch.LongTensor)

batch_size = 100
epochs = 30

train = TensorDataset(train_X, train_y)
val = TensorDataset(val_X, val_y)

train_loader = DataLoader(train, batch_size=batch_size, shuffle= False)
val_loader = DataLoader(val, batch_size=batch_size, shuffle=False)

````

````python
class ANNmodel(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim)
        self.fc3 = nn.Linear(hidden_dim, hidden_dim)
        self.fc4 = nn.Linear(hidden_dim, output_dim)

    def forward(self, X):
        a = self.fc1(X)
        a = F.relu(a)
        a = self.fc2(a)
        a = F.tanh(a)
        a = self.fc3(a)
        a = F.elu(a)
        out = self.fc4(a)
        return out

input_dim = 28 * 28
hidden_dim = 150
output_dim = 10

model = ANNmodel(input_dim, hidden_dim, output_dim)

error = nn.CrossEntropyLoss()

learning_rate = 0.02
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
````

````python
count = 0
train_loss_list = []
train_acc_list = []
val_loss_list =[]
val_acc_list = []
for epoch in range(epochs):
    total = 0
    correct = 0
    for images, labels in train_loader:
        X = Variable(images.reshape(-1, 28* 28))
        y = Variable(labels)

        optimizer.zero_grad()

        outputs = model(X)

        loss = error(outputs, y)

        loss.backward()

        optimizer.step()

        count += 1

        if count % 100 == 0:
            total += len(y)
            _, predicted = torch.max(outputs.data, axis=1)
            correct += (predicted == y).sum()
            acc = 100* (correct/total)
            print(f"iteration:{count}, loss:{loss.item()}, acc:{acc}%")

    train_loss_list.append(loss.item())
    train_acc_list.append(acc)

    val_loss = 0
    val_total = 0
    val_correct = 0
    with torch.no_grad():
        for images, labels in val_loader:
            val_total += len(labels)
            outputs = model(images)
            val_loss += error(outputs, labels).item()
            _, predicted = torch.max(outputs.data, axis=1)
            val_correct += (predicted == labels).sum()
    avg_val_loss = val_loss / len(val_loader)
    avg_val_acc = 100 * (val_correct / val_total)
    val_loss_list.append(avg_val_loss)
    val_acc_list.append(avg_val_acc)
    print(f"epoch{epoch + 1}, val_loss:{avg_val_loss}, val_acc:{avg_val_acc}%")
````

````python
plt.plot(range(epochs), train_loss_list, label='training loss', color='blue')
plt.plot(range(epochs), val_loss_list, label='validation loss', color='red')
plt.legend()
plt.xlabel("epoch")
plt.ylabel("loss")
plt.title("loss")
plt.show
````

````python
plt.plot(range(epochs), train_acc_list, label="training accuracy", color='blue')
plt.plot(range(epochs), val_acc_list, label="validation", color='red')
plt.legend()
plt.xlabel("epoch")
plt.ylabel("accuracy")
plt.title("accuracy")
plt.show()
````

###CNN

````python
class CNNmodel(nn.Module):
    def __init__(self):
        super().__init__()
        self.cnn1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=5, stride=1, padding=0)
        self.cnn2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=5, stride=1, padding=0)
        self.cf1 = nn.Linear(32*4*4, 10)

    def forward(self, X):
        a = self.cnn1(X)
        a = F.relu(a)
        a = F.max_pool2d(a, kernel_size=2)

        a = self.cnn2(a)
        a = F.relu(a)
        a = F.max_pool2d(a, 2)

        a = torch.flatten(a, start_dim=1) # 保留batch那個維度 從1維度開始攤平

        out = self.cf1(a)
        return out

model = CNNmodel().to(device) # 重新建立
learning_rate = 0.1

error = nn.CrossEntropyLoss().to(device)
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate) # 所有模型參數都在model.parameter()裡面np
````

````python
count = 0
train_loss_list = []
train_acc_list = []
val_loss_list =[]
val_acc_list = []
for epoch in range(epochs):
    total = 0
    correct = 0
    for images, labels in train_loader:
        X = Variable(images.reshape(-1, 1, 28, 28)).to(device) # CNN要確保輸入的batch, channel, width, height
        y = Variable(labels).to(device)

        optimizer.zero_grad()

        outputs = model(X)

        loss = error(outputs, y)

        loss.backward()

        optimizer.step()

        count += 1

        if count % 100 == 0:
            total += len(y)
            _, predicted = torch.max(outputs.data, axis=1)
            correct += (predicted == y).sum()
            acc = (100* (correct/total)).item()
            print(f"iteration:{count}, loss:{loss.item()}, acc:{acc}%")

        train_loss_list.append(loss.item())
        train_acc_list.append(acc)

    val_loss = 0
    val_total = 0
    val_correct = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)
            val_total += len(labels)
            outputs = model(images.reshape(-1, 1, 28, 28)) # CNN要確保輸入的batch, channel, width, height
            val_loss += error(outputs, labels).item()
            _, predicted = torch.max(outputs.data, axis=1)
            val_correct += (predicted == labels).sum()
    avg_val_loss = val_loss / len(val_loader)
    avg_val_acc = (100 * (val_correct / val_total)).item() # 因為有丟入device所以需要注意 GPU Tensor取值
    val_loss_list.append(avg_val_loss)
    val_acc_list.append(avg_val_acc)
    print(f"epoch{epoch + 1}, val_loss:{avg_val_loss}, val_acc:{avg_val_acc}%")
````
