#Pandas 基礎

```import pandas as pd```

### series

一維 本質為一個column

* ```df = pd.Series([list], index=[list])``` 建立Series

* ```df[index/'name']``` 透過index 或是 name 讀取column資料

### DataFrame

二維 由Series集合組成的多維表格

透過list建立

* ```df = pd.DataFrame([[list1], [list2]], index=[], columns=[])```

  一個list為一個row建立 index 與 columns 可省略

透過dict與list建立

* ```df = pd.DataFrame({'column1' : ['Bob', 'Amy'], 'column2' : [23, 25]}, index=[])```

  dict一個key:value為一個column建立

  index 可省略

透過list與dict建立

* ```df = pd.DataFrame([{'Name': 'Bob', 'Age': 23}, {'Name': 'Amy', 'Age': '25'}], index=[])```

  一個dict為一個row建立 key為column name

  index 可省略

取值 loc/iloc 是取row 純[]/.是取column 都可以透過輸入list一次抓取一群

* ```df.loc[index_name]``` 透過名稱或是索引讀取row Series type （索引[0,10]會包含10）

* ```df.iloc[index]``` 透過索引讀取row Series type （[0, 10]不包含10）

* ```df[column_name]``` 透過名稱讀取column Series type

* ```df.column_name``` 讀取column(少用)

* ```df[[column_name]]``` 透過名稱讀取column DataFrame type

* ```df.loc[:, name]``` 透過名稱使用切片讀取cloumn Series type

* ```df.iloc[:, column_index]``` 透過column索引使用切片讀取 column Series type

* ```df.loc[df['column_name1'] == 'value', ['column_name2', 'column_name3']]``` 透過判別式取出沒有index的DataFrame某個列1為某值的 列2跟列3的值
loc裡面也可以輸入bool的一個相同長度的list來選取

* ```df.loc[(df['column1'].isin(['arg1', 'arg2'])) & df.column2 >= 90]``` loc要透過bool取值時 []內的東西如果是不同column的名稱又多個時候要用df['column'].isin讀取

增加值

* ```df['column'] = 'arg'``` 建立一個新的column值都為arg

* ```df['column'] =  range(0,len(df))``` 可透過range填值 （len(df)會輸出row數量）

###groupby('') 可將df依照()中的column做分類 在取個別的統計指標

* 可接```df['column']``` ```.max()``` ```.min()``` ```mean()``` ```.apply()``` ```.map()```

* ```groupby(['column1', 'column2'])``` 可以分更多小群組

* 但分越多群組會需要越複雜的index讀取 透過 ```df.reset_index()```可以轉換為原本的index系統 但保留groupby分類的排序

* ```df.groupby('column1').column2.max()``` 舉例：對column1先分類 每個分類中的column2找最大值

* ```df.groupby('column1').column2.agg(['func1', 'func2'])``` 建立一個新的df透過column1分類後依照agg中 不包含()只給位址的func 建立columns


### I/O

* ```df = pd.read_csv('file')``` 讀取csv檔

* ```df = pd.read_excel('file')``` 讀取excel檔

* ```df = pd.read_json('file')``` 讀取json檔

* ```df = df.to_csv('file_name',index)``` 寫入csv index是否保留True放在新的一行False拋棄

* ```df.to_excel('file_name',sheet_name=,index)``` 寫入excel

* ```df.to_json('file_name')``` 寫入json

### 操作

* ```df.shape``` 取得行列數量 回傳Tuple

* ```df.to_string()``` 文件完整顯示

* ```df.head()``` 顯示前5個row 括號中可填入數字設定

* ```df.tail()``` 顯示後5個row

* ```df.describe()``` 分析統計指標  ```(percentiles=[0.3, 0.5,0.7])```

* ```df['column'].dtype 查詢該column的資料型態

* ```df.dtypes``` 查詢每個column分別是什麼資料型態

* ```df['column'].astype('')``` 改變某個column為()中的資料型態 (index也有資料型態 可以改變)

* ``````

* ```df.T``` 轉置

* ```df.info()``` 資料理解 幾個column 各名稱 多少非0 資料型態

* ```df.columns``` 取得所有column name

* ```df.isnull()/notnull()``` 檢查缺失值 回傳一個與df一樣大的DataFrame內容為bool

* ```df.isnull().sum()``` 列出每個column個別有多少個缺失值

* ```df.isnull().sum().sum()``` 計算全部總共多少個缺失值

* ```df.dropna(axis=0)``` 丟掉有缺失值的row

* ```df['column'].fillna('str')``` NaN會填入str

* ```df['column'].replace('str1', 'str2')``` str1換str2

* ```df[df.index.isin(range(2, 10))]``` 選擇特定範圍的row

* ```df['column'].count()``` 計算column有幾筆

* ```df['column'}.value_counts()``` 計算該column每個值出現幾次 有點set的概念加上次數

* ```df['column'].unique()``` 顯示column不重複的所有內容

* ```df['column'].map(lambda)``` 對該column所有row的值都做lambda（一般map是 arg=list(map(lambda,list))）

* ```df.apply(lambda,axis)``` axis=0/1 row/column 可以回傳一個新的df對全部值做lambda

* ```df['column1'] + df['column2']```相加兩個column的內容

* ```Series.idxmax()``` 可找到最大值的index

* ```df.sort_value(by='column')``` 整個df會依照column的大小由小排到大 (by鐘可以放一個list他會先按照前者慢慢往後順序排列) ()中放入ascending=False可以由大排到小

* ```df.sort_index()``` 透過index排序

* ```df.rename(column={'str1' : 'str2'}) 透過column/index更改欄位名稱

* ```df.rename_axis('str', axis='rows')``` 給row/column的名稱欄位一個名稱

* ```pd.concat([df1, df2])``` 兩個df上下對接

* ```df1.join(df2, lsuffex='str1', rsuffex='str2')``` 兩個df左右對接如果名稱重複需要後面兩個參數

* ```df.set_index('column')``` 將column改設為index
